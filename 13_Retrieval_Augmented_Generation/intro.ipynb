# Introduction to Retrieval Augmented Generation (RAG)

## 🚀 What is RAG?
**Retrieval Augmented Generation (RAG)** is a framework that combines **information retrieval** with **text generation** to enhance Large Language Models (LLMs). It allows LLMs to:
- Access **external knowledge** (beyond their parametric memory).
- Answer questions about **private data**, **recent events**, or niche domains.
- Reduce hallucinations by grounding responses in retrieved context.

---

## 🔍 Why RAG?
Traditional LLMs face limitations:
1. **Private Data**: Cannot answer questions about unseen/internal data.
2. **Temporal Knowledge**: Limited to pre-training cutoff dates.
3. **Hallucinations**: Generate plausible but incorrect answers.

RAG solves these by dynamically injecting **relevant context** into prompts.

---

## 🛠️ How RAG Works?
### 4-Step Workflow:
1. **Indexing**  
   - Convert documents into searchable chunks (text splitting + embeddings).
   - Store in a **vector database** (e.g., FAISS, Pinecone).

2. **Retrieval**  
   - Use semantic search to fetch top-`k` chunks relevant to the user query.

3. **Augmentation**  
   - Inject retrieved context into the LLM prompt:
   ```python
   prompt = f"""
   Answer using ONLY this context:
   {retrieved_text}
   
   Question: {user_query}
   """
4. **Generation**
   - LLM generates answers conditioned on the retrieved context.

## 🆚 RAG vs Fine-Tuning

| Aspect         | RAG                                | Fine-Tuning                        |
|----------------|-------------------------------------|------------------------------------|
| **Cost**       | Low (no model retraining)           | High (compute/resources)           |
| **Data Updates** | Real-time (swap context)            | Requires retraining                |
| **Hallucinations** | Reduced (grounded in context)      | Depends on training data           |
| **Use Case**   | Dynamic/private knowledge           | Domain-specific task mastery       |
